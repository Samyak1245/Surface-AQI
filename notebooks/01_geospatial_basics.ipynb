```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

print("Environment is working successfully")
```

```python
data = {
    "city": ["Delhi", "Mumbai", "Pune", "Kolkata", "Bengaluru"],
    "latitude": [28.61, 19.07, 18.52, 22.57, 12.97],
    "longitude": [77.23, 72.87, 73.85, 88.36, 77.59],
    "pm25": [180, 75, 65, 120, 45]
}

df = pd.DataFrame(data)
df
```

# Convert the table into geographic points

```python
gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(
        df["longitude"],
        df["latitude"]
    ),
    crs="EPSG:4326"
)

gdf
```

# Plot the locations

```python
gdf.plot(
    figsize=(8, 6),
    markersize=100,
    column="pm25",
    legend=True
)

plt.title("Sample PM2.5 Values")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```
